# Atividade Prática - Otimização de Rotas de Vistorias Rodoviárias

## 1. Contexto do problema


Um órgão de fiscalização rodoviária do estado do Pará precisa planejar uma rota de vistorias para percorrer quinze pontos localizados nas rodovias federais do estado. Cada ponto representa uma posição intermediária aproximada de cada rodovia e foi previamente identificado em um GeoDataFrame, de modo que juntos cobrem todo o território paraense.

A tarefa consiste em determinar a **melhor rota possível** que permita visitar todos os 15 pontos de vistoria **iniciando e retornando à capital Belém**, minimizando a distância total percorrida.

## 2. Bases de dados

Você receberá dois arquivos com informações sobre os pontos a serem visitados.

- `brs-pa-coordinates.pkl`: arquivo *pickle* com um `GeoDataFrame` contendo:

    - `identificador`: código da rodovia federal (ex: BR-010, BR-153);
    - `geometry`: geometria da rodovia (objeto LINESTRING);
    - `coordinate`: ponto médio da rodovia em coordenadas geográficas (longitude, latitude).

- `brs-pa-distances.csv`: matriz de distâncias par-a-par entre os 15 pontos de vistoria (em quilômetros), usada como entrada para o cálculo do custo da rota.

A partir desses dados, o grupo deve:

1. Definir o ponto inicial mais próximo de Belém;
2. Utilizar a matriz de distâncias para calcular o percurso total das soluções geradas;
3. Implementar e ajustar os parâmetros do ACO para convergir à melhor rota.

## 3. Restrições obrigatórias

A sua solução  e implementação deve respeitar as seguintes regras:

- O algoritmo deve **visitar todos os 15 pontos** uma única vez e **retornar ao ponto inicial**.
- O grupo deve **utilizar o ACO clássico**, com estrutura baseada em **feromônio, visibilidade e evaporação**.
- O cálculo de custo deve ser feito **com base na matriz `brs-pa-distances.csv`** (não em coordenadas geográficas).
- O relatório deve incluir um **gráfico ou mapa da rota final** (usando *folium*, *matplotlib* ou *geopandas*).
- Os experimentos devem explorar **ao menos 3 variações de parâmetros** (ex: número de formigas, taxa de evaporação, α, β).

## 4. Entrega

A entrega da atividade deve ser feita em um **repositório Git do grupo**, contendo os seguintes elementos organizados de forma clara e padronizada:

### Estrutura do repositório

- `notebook.ipynb`: **Notebook principal** com:

  - explicações em *markdown* sobre o **contexto do problema**, a **modelagem** e a **estratégia adotada**;
  - trechos principais de código executável (como chamadas de funções, gráficos, análise de resultados).

- `src/`: **Código-fonte modularizado** com:

  - organização em **funções e/ou classes reutilizáveis**;
  - separação clara entre função objetivo, inicialização, algoritmo de busca, visualização, etc.

- `data/`: Arquivos de dados utilizados (.csv, .pkl, etc).

- `figures/`: Imagens e gráficos gerados, especialmente os de **convergência** e **comparação de resultados**.

- `README.md`: Arquivo com:

  * 📌 **Contexto do problema**;
  * ⚙️ **Instruções de instalação e execução** do projeto;
  * 💡 **Resumo da solução implementada**;
  * 📊 **Descrição da experimentação** (configurações e análise dos resultados).


### Requisitos de experimentação

Todos os algoritmos devem ser **executados 5 vezes**, e o relatório final deve incluir:

- **Melhor, pior, média e desvio padrão** das soluções obtidas;
- **Tempo médio de execução** e seu **desvio padrão**;
- **Gráfico de convergência** (fitness por iteração) para pelo menos uma das execuções;
- Análise interpretativa dos resultados e possíveis melhorias.

### Específico ao ACO

- Implementar a **função objetivo** que calcula a distância total de uma rota entre os 15 pontos rodoviários.
- Utilizar o arquivo `brs-pa-distances.csv` com as distâncias par-a-par e coordenadas dos pontos.
- Implementar o algoritmo ACO clássico com atualização de feromônio, evaporação e deposição de feromônio.
- Exibir a **melhor rota** encontrada (sequência dos pontos), com a **distância total** e gráfico de convergência.
- Realizar 5 execuções para análise estatística dos resultados.
