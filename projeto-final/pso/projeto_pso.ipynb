# Atividade Prática - Otimização de Componentes de Concreto

## 1. Contexto do problema


Concreto é um dos materiais mais utilizados na engenharia civil, e sua **resistência à compressão** depende diretamente da proporção dos materiais usados em sua composição. O objetivo desta atividade é determinar a melhor combinação de componentes do concreto para **maximizar sua resistência à compressão aos 28 dias**, utilizando o algoritmo de **Otimização por Enxame de Partículas (PSO)**.

Para isso, vocês usarão um modelo preditivo (Random Forest) já treinado com dados reais, que estima a resistência do concreto com base nas quantidades (em kg/m³) dos sete componentes principais do material.

## 2. Bases de dados

Os seguintes arquivos serão fornecidos:

- `concrete_data.csv`: contém o [conjunto de dados original](https://archive.ics.uci.edu/dataset/165/concrete+compressive+strength) com 1030 amostras de composições de concreto e suas respectivas resistências à compressão.
- `rf_concrete.pkl`: modelo Random Forest treinado com esses dados, capaz de prever a resistência com boa precisão.
- `concrete_predictor.py`: arquivo com a classe `ConcreteStrengthPredictor`, que carrega o modelo e fornece um método `predict()` para estimar a resistência com base nas sete variáveis de entrada.

O dataset original vai ser usado como referência dos valores dos componentes do concreto, e é composto pelos seguintes dados:

```python
import pandas as pd

df = pd.read_csv("concrete_data.csv")
df
```

A classe `ConcreteStrengthPredictor` pode ser utilizada da seguinte forma:

```python
from concrete_predictor import ConcreteStrengthPredictor

predictor = ConcreteStrengthPredictor()

strength = predictor.predict(
    cement=400,
    blastFurnaceSlag=100,
    flyAsh=0,
    water=150,
    superplasticizer=10,
    coarseAggregate=1000,
    fineAggregate=800
)

print(f"Predicted strength: {strength:.2f} MPa")
```

## 3. Restrições obrigatórias

A sua solução e implementação deve respeitar as seguintes regras:

- **Utilizar a classe `ConcreteStrengthPredictor`** contida em `concrete_predictor.py` para fazer previsões. **Não é necessário treinar o modelo novamente.**
- Implementar uma **função objetivo** para o PSO que:

   - Recebe as 7 variáveis de entrada: `cement`, `blastFurnaceSlag`, `flyAsh`, `water`, `superplasticizer`, `coarseAggregate`, `fineAggregate`.
   - Retorna a **força prevista**, penalizada por desvio da soma das massas em relação à média (valor de referência):
   $$J(x) = 
   \begin{cases}
   f(x_1, ..., x_7, 28) & , \text{se } \sum_{i=0}^{6}x_{i} \leq M\\
   f(x_1, ..., x_7, 28) - \lambda (\sum_{i=1}^{7}x_{i} - M)^{2} & ,\text{se } \sum_{i=0}^{6}x_{i} > M
   \end{cases}
   $$
     Onde:
     - $f(x_1, ..., x_7, 28)$ é a força prevista pelo modelo de Random Forest.
     - $\lambda = 0,0002$ é um coeficiente de penalização de soluções cuja soma dos valores ultrapassa $M$.
     - $M$ é o valor médio da soma dos componentes no dataset ($\approx$ 2343,5 kg/m³).
3. Implementar e aplicar o algoritmo **canônico de PSO** para encontrar a melhor combinação dos 7 valores contínuos.
4. Respeitar os **limites mínimo e máximo** de cada variável conforme observado no `concrete_data.csv`.
